```python
#input numbers and operand
first_number = float(input("Kindly enter the first number:"))
operand = input("please enter the operand of your choice, (+,-,/,*)")
second_number = float(input("Kindly enter the second number:"))
```

```python
#check for only operators
valid_operators = ["+", "-", "/", "*"]
if operand not in valid_operators:
    print("this is not an operator, please input and operator")
    exit() 
```

```python
#where the operand actually work
try:
    if operand == "*":
        print("your answer is,", first_number * second_number)
        elif operand == "/":
            print("your answer is,", first_number / second_number)
    if second_number == "0":
        print("cannot divide by 0, input an actual number"):
            else print("your answer is:", first_number/second_number)
    if operand == "+":
        print("your answer is,", first_number + second_number)
        elif operand == "-":
            print("your answer is," first_number - second_number)

except ValueError:
    print("invalid number, input an actual number")
```
