```python
#STUDENT PERFORMANCE PREDICTOR
import pandas as pd
from sklearn.tree import DecisionTreeClassifier

import matplotlib
matplotlib.use("TkAgg")  # ensures plots open in a window (for scripts)
```

```python
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier, plot_tree 
%matplotlib inline

import tkinter as tk
from tkinter import messagebox
```

```python
#Create Dataset
data = {
    "Study_Hours": [1, 2, 3, 4, 5, 6, 2, 7, 8, 3, 9, 1, 5, 6, 2],
    "Attendance": [40, 50, 60, 70, 80, 90, 45, 85, 95, 55, 98, 30, 75, 88, 35],
    "Assignment_Score": [30, 40, 50, 60, 70, 85, 35, 90, 92, 45, 95, 25, 68, 80, 20],
    "Result": ["Fail", "Fail", "Fail", "Pass", "Pass", "Pass",
               "Fail", "Pass", "Pass", "Fail", "Pass", "Fail",
               "Pass", "Pass", "Fail"]
}
```

```python
#Put data in Data Frame
df = pd.DataFrame(data)
print(df) 
```

```python
X = df[["Study_Hours", "Attendance", "Assignment_Score"]]
y = df["Result"]
```

```python
#Train Model
model = DecisionTreeClassifier()
model.fit(X, y)

print("Model trained successfully!\n")
```

```python
#Take User input
Study_Hours = float(input("Enter study hours per day: "))
Attendance = float(input("Enter attendance percentage: "))
Assignment_Score = float(input("Enter assignment score: "))
```

```python
prediction = model.predict([[Study_Hours, Attendance, Assignment_Score]])

print("\nPrediction:", prediction[0])
```
