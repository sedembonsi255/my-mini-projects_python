```python
import pandas as pd
```

```python
file1 = pd.read_excel("D:\Desktop\sed_eeeem\FECESA\Telecommunication_Contacts_NoDepartment.xlsx" , engine="openpyxl")  # contains Name, Contact
file2 = pd.read_excel("D:\Desktop\sed_eeeem\FECESA\Extracted_Full_Names.xlsx", engine="openpyxl")  # contains Name
```

```python
print(file1.head())
print(file2.head())
```

```python
import numpy as np

n_contacts = len(file1)
n_names = len(file2)

# Compute size of each chunk
chunk_size = n_names // n_contacts
remainder = n_names % n_contacts

# Create list to hold assigned names
assigned_names = []

start = 0
for i in range(n_contacts):
    end = start + chunk_size
    # Distribute the remainder one by one
    if i < remainder:
        end += 1
    assigned_names.append(file2.iloc[start:end].copy())
    start = end
```

```python
# Build a new DataFrame for output
output_rows = []

for i, chunk in enumerate(assigned_names):
    contact_name = file1.iloc[i]['Full Name']
    contact_number = file1.iloc[i]['WhatsApp Number']
    for name in chunk['Full Name']:  # assuming File 2 column is 'Name'
        output_rows.append({
            'Contact_Name': contact_name,
            'Contact': contact_number,
            'Assigned_Name': name
        })

output_df = pd.DataFrame(output_rows)
```

```python
output_df.to_excel(r"D:\Desktop\sed_eeeem\FECESA\paired_output.xlsx", index=False)
```
